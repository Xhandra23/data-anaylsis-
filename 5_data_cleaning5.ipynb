```python
import pandas as pd
empl  = pd.DataFrame({ 'empid': [101,102,103,104],
                      'name' : ["asha",None,"menna","sai"],
                       'dept' : ["sales",'it',"hr",None],
                       "salry": [45000,60000,52000,89000]})
'''print(empl)'''
print(empl.head(2))
print(empl.tail(2))
print(empl[['name','salry']])
print(empl.iloc[2])
print(empl[empl["salry"] > 50000])
print(empl[empl["dept"] == "it"])
print(empl[(empl['name'] == None) | (empl["dept"]==None)] )

print(empl.isna().sum())

empl['name'].fillna("Unknow")
empl["dept"].fillna("Not assigned")

empl["Bonus"] = empl["salry"]*10/100
print(empl)

empl["salry"].where(empl['name'] != 'sai',95000)

'''empl.rename(columns ={'empid':"employeeID"},inplace= True )'''
print(empl)

empl.drop("Bonus",axis=1)

empl["salry"].mean()
empl['salry'].median()
empl["salry"].min()
empl['salry'].max()
empl['salry'].sum()


emp_d = pd.DataFrame({
    "empid":[101,102,103,104],
    "city":["hyderbad","chennai","bangalor","pune"],
    "experience":[2,3,4,5]
})
print(emp_d)
pd.merge(empl,emp_d,on="empid",how='outer')


new = pd.DataFrame({
    "empid"  : [105,106],
    "name":["sun","moon"]
})

pd.concat([empl,new,emp_d],axis=0,ignore_index=True)
```
