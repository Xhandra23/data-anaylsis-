```python
import matplotlib.pyplot as plt
import numpy as np
arr  = np.array(["laptop","Mobile","tablet","headphone","watch"])
sales =np.array([120,180,90,150,110])
profit=np.array([30,50,20,40,25])
print(arr)
print(sales)
print(profit)
plt.plot(arr,sales,marker="*",)
plt.title("product sales")
plt.xlabel("arr")
plt.ylabel("sales")
plt.show
```

```python
plt.plot(arr,sales,marker="o",label="sales")
plt.plot(arr,profit,marker="o",label="profit")
plt.legend()
```

```python
plt.bar(arr,sales)
plt.title("product sales")
plt.xlabel("arr")
plt.ylabel("sales")
plt.show
```

```python
plt.barh(arr,sales)
plt.title("products values")
plt.xlabel("sales")
plt.ylabel("arr")
plt.show
```

```python
plt.barh(arr,profit,color="black")
plt.title("products values")
plt.xlabel("porfit")
plt.ylabel("arr")
plt.show
```

```python
x=np.arange(len(arr))

width = 0.35
plt.bar(x+width/2,sales,width,label="sales",color="yellow")
plt.bar(x-width/2,profit,width,label="profit",color="black")
plt.xticks(x,arr)
plt.title("sales vs profit")
plt.xlabel("product")
plt.ylabel("amout")
plt.legend()
plt.show()
print(x)
```

```python
plt.pie(sales,labels=arr,autopct="%1.1f%%"
        )
plt.title("jhfnd")
plt.show()
```

```python

plt.pie(sales,labels=arr,autopct="%1.1f%%",wedgeprops={"width":0.2}
        )
plt.title("jhfnd")
plt.show()
```
