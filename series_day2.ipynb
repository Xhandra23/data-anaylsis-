```python
 import pandas as pd

weather = pd.Series(
    [25, 23, None, 30, 40, None, None, 36, 32, 20, None, 10],
    index=[
        "January", "February", "March", "April",
        "May", "June", "July", "August",
        "September", "October", "November", "December"
    ],
    name="Weather 2025"
)


''' vewing'''
weather.head()
weather.tail()

weather.shape
weather.index
len(weather)
weather.values
''' filtering'''
print(weather[[2,3,7,5]])
weather[weather > 10]

'''finding missing values'''
weather.isna()
print(weather.isnull().sum())
weather.dropna()
weather.fillna(0)
weather.fillna(weather.mean())
```

```python
''' aggregation functions'''
''' mean()
median()
sum()
min()
max()
std()
'''
import pandas as pd

sales = pd.Series(
    [5000, 6500, 4200, 7800, 5500, 9000, 7200, 6800, 8100, 4500],
    index=[
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday",
        "Saturday", "Sunday", "Week2-Mon", "Week2-Tue", "Week2-Wed"
    ]
)

print(sales)
print(sales.mean())
print(sales.median())
print(sales.sum())
print(sales.min())
print(sales.max())
print(sales.std())
```
