```python
import pandas as pd

data = {
    "Weekday": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"],
    "Sale": [1200, None, 900, 1800, None, 1100, 2500],
    "Profit": [200, 300, None, 450, 600, None, 700],
    "Loss": [None, 0, 100, None, 0, 50, 0],
    "Discount": [50, None, 75, 150, 200, 80, None]
}

df = pd.DataFrame(data)

df.isna()
df.isna().sum()

df.fillna(0)
df.fillna({ 'Sale':100,
           'Loss':10,
            'Discount': 45,
            'Profit' : 20})

df['Sale'] = df['Sale'].fillna(df['Sale'].mean())
print(df['Sale'])


''' create a new column in dataframe
df['id'] = [23,4,52,12,32,32,12]'''

''' create a new column useing exixting column'''
df['next'] = df['Sale']*10
print(df)

df.loc[df['Sale'] == 1800,'Profit'] = 1000000
print(df)
```
