```python

import pandas as pd
import numpy as np
date = pd.date_range(start="2023-09-23",periods=90,freq='D')
sales = np.random.randint(20000,30000,size=90)
data ={"Date":date,
       "Sales":sales}
df = pd.DataFrame(data)
print(df)

df.dtypes
df["Date"].dt.year
df["year"] =df["Date"].dt.year
print(df)

df.groupby(df["Date"].dt.year)["Sales"].sum()
df.groupby(df["Date"].dt.year)["Sales"].mean()
df.groupby(df["Date"].dt.year)["Sales"].max()
df.groupby(df["Date"].dt.year)["Sales"].sum()
```
