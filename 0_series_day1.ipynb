```python
''' create series in dictionary'''
import pandas as pd
a ={1:23,2:45,3:67}
b=pd.Series(a)
print(b)
''' updating sereis'''
b[2]= 54
print(b)
''' adding new value'''
b[4]=76
print(b)
'''remove a value'''
e= b.drop(1)
print(e)
'''insert bulk of values'''
new_data= pd.Series([23,4,5],index=[5,6,7])
f=pd.concat([b,new_data])
print(f)

''' delete bulk of of values '''
g= f.drop(2,4,6)
print(g)
```

```python
''' create series in dictionary'''
import pandas as pd
std_data ={101:78,102:85,103:92,104:67,105:88}
s = pd.Series(std_data)
print(s)

''' diaplay the marks of student 103'''
print(s[103])

''' change the marks of std 104 from 67 to 75 '''
s[104] = 75
print(s[104])

''' delete student 102 from the series'''
f = s.drop(102)
print(f)

''' add  a new student 106 and marks 90'''
s[106] = 90
print(s)
```
