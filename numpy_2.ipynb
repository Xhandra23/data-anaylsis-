```python
#ravel() it 2d array into 1d array
import numpy as np
ii = np.arange(1,100).reshape((1,9,11))
print(ii)
print(ii.ravel())
```

```python
#(.T) transpose matrixs
import numpy as np
arr = np.arange(1,19).reshape(6,3)
print(arr)
arr.T
```
