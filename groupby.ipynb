```python
import pandas as pd

data = {
    "EMPID": [101,102,103,104,105,106,107,108,109,110,110],

    "Name": ["Ravi","PRIYA","Arun","sneha","Kiran","ANJALI",
             "Vikram","meena","RAHUL","Sneha","Sneha"],

    "City": ["Hyderabad","Hyderabad","CHENNAI","Chennai","Bangalore",
             "BANGALORE","Mumbai","mumbai","Delhi","Hyderabad","Hyderabad"],

    "Salary": [30000,40000,3500,4500,32000,50000,
               38000,42000,36000,45000,45000],

    "Age": [25,28,24,26,30,27,32,29,31,23,23]
}

df = pd.DataFrame(data)

print(df)
df=pd.DataFrame(data)
print(df)
df.groupby('City')['Salary'].sum()
df.groupby('City')['Salary'].mean()
df.groupby('City')['Salary'].max()
df.groupby('City')['Salary'].agg(['sum','max','min','mean'])
df.groupby('City').agg({ 'Age' : ['sum','max'],
                                  'Salary' :['min','mean']})
```
